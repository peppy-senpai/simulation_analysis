```python
import random
import math

def Failure():
    global S
    global Slast
    global Tlast
    global Area
    global NextFailure
    global NextRepair
    
    S = S - 1
    if S >= 1:
        NextFailure = Clock + math.ceil(6*random.random())
    if S == 2:
        NextRepair = Clock + 3.5
    
    Area = Area + Slast * (Clock - Tlast)
    Tlast = Clock
    Slast = S
    
def Repair():
    global S
    global Slast
    global Tlast
    global Area
    global NextFailure
    global NextRepair
    
    S = S + 1
    
    if S == 2:
        NextRepair = Clock + 3.5
    if S == 1:
        NextFailure = Clock + math.ceil(6*random.random())
    
    Area = Area + Slast * (Clock - Tlast)
    Tlast = Clock
    Slast = S
    
def Timer():
    global Clock
    global NextFailure
    global NextRepair
    
    if NextFailure < NextRepair:
        result = "Failure"
        Clock = NextFailure
        NextFailure = Infinity
    else:
        result = "Repair"
        Clock = NextRepair
        NextRepair = Infinity
    return result

# Print detailed sample path for one simulation
def print_sample_path():
    print("\n" + "="*80)
    print("DETAILED SAMPLE PATH FOR SINGLE SIMULATION RUN")
    print("="*80)
    print(f"{'Event':<10} {'Clock':<10} {'S':<5} {'NextFail':<12} {'NextRepair':<12} {'Area':<10}")
    print("-"*80)

Infinity = float('inf')
random.seed(1234)

# Run one detailed simulation first
print_sample_path()
NextFailure = math.ceil(6*random.random())
NextRepair = Infinity

S = 3.0
Slast = 3.0
Clock = 0.0
Tlast = 0.0
Area = 0.0

print(f"{'Start':<10} {Clock:<10.2f} {int(S):<5} {NextFailure:<12.2f} {'Inf':<12} {Area:<10.2f}")

while S > 0:
    NextEvent = Timer()
    if NextEvent == "Failure":
        Failure()
    else:
        Repair()
    
    nf_str = f"{NextFailure:.2f}" if NextFailure != Infinity else "Inf"
    nr_str = f"{NextRepair:.2f}" if NextRepair != Infinity else "Inf"
    print(f"{NextEvent:<10} {Clock:<10.2f} {int(S):<5} {nf_str:<12} {nr_str:<12} {Area:<10.2f}")

print("-"*80)
print(f"System failed at time: {Clock:.2f}")
print(f"Average number of functional components: {Area/Clock:.4f}")
print("="*80)

# Run 100 replications for statistics
SumS = 0
SumY = 0
for Reps in range(1):
    NextFailure = math.ceil(6*random.random())
    NextRepair = Infinity

    S = 3.0
    Slast = 3.0
    Clock = 0.0
    Tlast = 0.0
    Area = 0.0 
    while S > 0:
        NextEvent = Timer()
        if NextEvent == "Failure":
            Failure()
        else:
            Repair()
            
    SumS = SumS + Area/Clock
    SumY = SumY + Clock

print("\n" + "="*80)
print("RESULTS FROM 100 REPLICATIONS")
print("="*80)
print("Average failure time: " + str(SumY/100) +
 " days\nAverage # of functional components: " + str(SumS/100))
print("="*80)
```
